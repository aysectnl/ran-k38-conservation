```python
import os

hedef_dosya = "all.aln.fasta"
bulunan_yol = None

# Aramayı en üst dizinden (/) başlatıyoruz ki var, usr, sys nerede olursa olsun bulsun
for root, dirs, files in os.walk('/'):
    if hedef_dosya in files:
        bulunan_yol = os.path.join(root, hedef_dosya)
        break

if bulunan_yol is None:
    print(f"Hata: {hedef_dosya} hiçbir klasörde bulunamadı.")
else:
    print(f"✅ Dosya başarıyla bulundu: {bulunan_yol}\n")
    seqs = {}
    with open(bulunan_yol, 'r') as f:
        header = ""
        for line in f:
            line = line.strip()
            if line.startswith('>'):
                header = line[1:]
                seqs[header] = ""
            else:
                seqs[header] += line

    human_header = None
    for key in seqs.keys():
        if 'RAN_HUMAN' in key:
            human_header = key
            break

    if not human_header:
        print("Hata: RAN_HUMAN dizisi dosyada bulunamadı.")
    else:
        human_seq = seqs[human_header]
        res_count = 0
        target_col = -1

        for idx, char in enumerate(human_seq):
            if char != '-':
                res_count += 1
            if res_count == 38:
                target_col = idx
                break

        print(f"RAN_HUMAN K38 için hizalama kolonu (1-tabanlı): {target_col + 1}")
        print("-" * 50)

        for name, seq in seqs.items():
            if target_col < len(seq):
                kisa_isim = name.split('|')[-1].split()[0] if '|' in name else name.split()[0]
                print(f"{kisa_isim:<15}: {seq[target_col]}")
```
